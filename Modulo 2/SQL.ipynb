```python
df = pd.read_csv("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/labs/module_2/data/Spacex.csv")
df.to_sql("SPACEXTBL", con, if_exists='replace', index=False,method="multi")
```

```python
%sql DROP TABLE IF EXISTS SPACEXTABLE;
```

```python
%sql create table SPACEXTABLE as select * from SPACEXTBL where Date is not null
```

```python
##Mostrar los nombres de los sitios de lanzamiento únicos en la misión espacial
%sql  select distinct Launch_Site from SPACEXTABLE ;
```

```python
## Mostrar 5 registros donde comienzan los sitios de lanzamiento con la cadena 'CCA' 
%sql select * from SPACEXTABLE  WHERE Launch_Site LIKE 'CCA%' LIMIT 3;
```

```python
## Tarea 3: Mostrar la masa total de carga útil transportada por los propulsores lanzados por la NASA (CRS)
```

```python
%sql select sum(PAYLOAD_MASS__KG_) as 'Total Masa  ' from SPACEXTABLE where  Customer like 'NASA (CRS)%' 
```

```python
## Tarea 4: Mostrar la masa media de carga útil transportada por el propulsor versión F9 v1.1
%sql select avg(PAYLOAD_MASS__KG_)  as 'total media  de Masa'from SPACEXTABLE where Booster_Version like 'F9 v1.1%'
```

```python
## Tarea 5: Enumera la fecha en que se logró el primer resultado exitoso de aterrizaje en la plataforma terrestre
%sql select min(date) as  'fecha primer resultado 'from SPACEXTABLE where Landing_Outcome like 'Success (ground pad)';
```

```python
##Tarea 6: Enumera los nombres de los propulsores que tienen éxito en naves dron y que tienen una carga útil 
##superior a 4000 pero inferior a 6000
%sql select  Booster_Version from SPACEXTABLE where Landing_Outcome like 'Success (drone ship)' and PAYLOAD_MASS__KG_  between 4000 and 6000;
```

```python
##Tarea 7: Enumera el número total de resultados exitosos y fallidos de misiones
%sql select  Mission_Outcome  , count(*)  from SPACEXTABLE  group by Mission_Outcome; 
```

```python
## Tarea 8: Enumera todos los booster_versions que han soportado la masa máxima de carga útil, usando una subconsulta
%sql select DISTINCT Booster_Version from  SPACEXTABLE WHERE PAYLOAD_MASS__KG_ ==  15600;
```

```python
##Tarea 9: Enumera los registros con el mes, landing_outcomes de fallo en naves dron, 
##booster_version y launch_site para el año 2015

%sql SELECT SUBSTR(Date, 6, 2) AS Month, Landing_Outcome, Booster_Version, Launch_Site FROM SPACEXTABLE WHERE SUBSTR(Date, 1, 4) = '2015' AND Landing_Outcome = 'Failure (drone ship)';
```

```python
## Tarea 10: Clasifica el conteo de resultados de aterrizaje entre 2010-06-04 y 2017-03-20, en orden descendente
%sql SELECT Landing_Outcome, COUNT(*) AS Count FROM SPACEXTABLE WHERE Date BETWEEN '2010-06-04' AND '2017-03-20' GROUP BY Landing_Outcome ORDER BY Count DESC;
```
